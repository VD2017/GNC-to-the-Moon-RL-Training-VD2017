```python
import redis



redis_client = redis.Redis(host= "192.168.50.32", port= 6379, decode_responses=False)
# redis_client = redis.Redis(host= "localhost", port= 6379, decode_responses=True)

redis_client.set('foo', 'bar')

print(redis_client.get('foo'))


```

```python
# redis stored Tensor 
import pickle
import torch

test_tensor = torch.ones((10))

tensor_key = "MyTensor"

pickled_tensor = pickle.dumps(test_tensor)
# redis_client.
# print(pickled_tensor)
redis_client.set(tensor_key, pickled_tensor)
# print(redis.StrictRedis())
retreived_data = redis_client.get("MyTensor")
reloaded_tensor = pickle.loads(retreived_data)
print(type(reloaded_tensor))
```
